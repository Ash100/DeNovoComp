DeNovoComp lets you **design new drug-like molecules to a target property profile**,
not just sample random structures. You declare what you want (molecular weight,
lipophilicity, polar surface area, a scaffold to keep, a lead to stay similar to,
toxicophores to avoid), that declaration is compiled into a reward, and a deep
generative model is ***steered*** toward molecules that satisfy it by reinforcement
learning (RL).

The survivors are then filtered, scored for **synthesizability**, checked for
**structural alerts**, tested against the **applicability domain**, and triaged into a
ranked shortlist ready to hand to the ADMET pipelines.

> **This notebook reproduces every figure and table of the DeNovoComp manuscript.**
> It runs **multiple design campaigns in one execution**, measures the
> **compliance-versus-diversity frontier** across reinforcement-learning intensities,
> and renders all display items at **600 dpi**.

# **Part 2: Training Data** - teaching the model *what a molecule looks like*

The generator first has to learn the "grammar" of real chemistry from a set of example
molecules. Here you only choose the *chemical universe* the model draws its vocabulary
and instincts from.

Two options:

1. **Default**: a curated drug-like set (ZINC-250k) is downloaded automatically. Good
   general-purpose starting point.
2. **Your own CSV**: set `USER_CSV_PATH` to a file with a `smiles` column. Use this to
   bias the model toward a chemical series you care about.

### **For Your Read, Cell 2.1: Load, clean, standardise and convert to SELFIES**

**What you're doing:** turning a messy list of SMILES into a clean, deduplicated,
model-ready dataset.

**What happens to each molecule:**
1. Parsed and sanitised by RDKit; anything unparseable is dropped.
2. Canonicalised, so duplicates written differently collapse to one entry. This step
   comes *before* deduplication for a reason: one molecular graph admits many valid
   SMILES strings, and without canonicalisation the same molecule would be presented
   repeatedly as independent training evidence.
3. Size-filtered to 5-50 heavy atoms, which removes salts, single atoms and outliers.
4. Converted to SELFIES and length-capped at `MAX_LEN` tokens.

**What to check in the output:** the printed counts tell you how many molecules survived
each step. If your custom CSV loses most of its rows here, investigate before training.

````python
#@title Cell 2.1:  Load + standardise + SELFIES

USER_CSV_PATH = None    # e.g. "/content/my_actives.csv"

ZINC_URLS = [
    "https://raw.githubusercontent.com/aspuru-guzik-group/chemical_vae/main/"
    "models/zinc_properties/250k_rndm_zinc_drugs_clean_3.csv",
    "https://raw.githubusercontent.com/aspuru-guzik-group/chemical_vae/master/"
    "models/zinc_properties/250k_rndm_zinc_drugs_clean_3.csv",
]

def _try_download(urls, dest):
    import urllib.request
    for url in urls:
        try:
            print(f"  trying {url[:70]}\u2026")
            urllib.request.urlretrieve(url, dest)
            if Path(dest).stat().st_size > 1024:
                return True
        except Exception as e:
            print(f"    failed: {e}")
    return False

def load_training_smiles():
    if USER_CSV_PATH and Path(USER_CSV_PATH).exists():
        print(f"\U0001f4c4 Loading user CSV: {USER_CSV_PATH}")
        df = pd.read_csv(USER_CSV_PATH)
        col = next((c for c in df.columns
                    if c.lower() in ("smiles", "smi", "canonical_smiles")), None)
        if col is None:
            raise ValueError("No SMILES column found (smiles / smi / canonical_smiles).")
        return [str(s).strip() for s in df[col]], "user_csv"

    cache = OUT / "zinc_250k.csv"
    if not cache.exists():
        print("\U0001f310 Downloading ZINC-250k (default training set)\u2026")
        if not _try_download(ZINC_URLS, cache):
            print("\u26a0\ufe0f  Network fallback \u2192 tiny embedded seed set.")
            seed = ["CCO", "CC(=O)O", "c1ccccc1", "CCN(CC)CC",
                    "CC(C)Cc1ccc(cc1)C(C)C(=O)O",
                    "CC(C)NCC(O)COc1ccc(cc1)CC(=O)N", "O=C(C)Oc1ccccc1C(=O)O",
                    "CN1C=NC2=C1C(=O)N(C(=O)N2C)C", "CC(=O)Nc1ccc(O)cc1",
                    "CN(C)CCC=C1c2ccccc2CCc2ccccc21", "COc1ccc2nc(S(N)(=O)=O)sc2c1",
                    "CC(C)NC(=O)C1=CC=CC=C1O", "COC(=O)C1=CC=CC=C1O",
                    "CC(C)(C)NCC(O)COc1ccc(cc1)CC(N)=O"] * 800
            pd.DataFrame({"smiles": seed}).to_csv(cache, index=False)

    df = pd.read_csv(cache)
    col = next(c for c in df.columns
               if c.lower() in ("smiles", "smi", "canonical_smiles"))
    return [str(s).strip() for s in df[col]], "zinc_250k"

raw_smiles, source_name = load_training_smiles()
if CFG.get("MAX_RAW"):
    raw_smiles = raw_smiles[:CFG["MAX_RAW"]]
print(f"\u2713 Loaded {len(raw_smiles):,} SMILES from source: {source_name}")

def _canon(smi):
    try:
        m = Chem.MolFromSmiles(smi)
        if m is None or not (5 <= m.GetNumHeavyAtoms() <= 50):
            return None
        return Chem.MolToSmiles(m, canonical=True)
    except Exception:
        return None

canon = [_canon(s) for s in tqdm(raw_smiles, desc="Canonicalising")]
n_dropped = sum(1 for c in canon if c is None)
canon = [s for s in canon if s]
n_before_dedup = len(canon)
canon = list(dict.fromkeys(canon))                 # dedup, keep order
print(f"  parsed / size-filtered : {n_before_dedup:,}  (dropped {n_dropped:,})")
print(f"  after deduplication    : {len(canon):,}")

tok_lens, pairs = [], []
for s in tqdm(canon, desc="SMILES \u2192 SELFIES"):
    try:
        ss = sf.encoder(s)
    except Exception:
        ss = None
    if not ss:
        continue
    L = sf.len_selfies(ss)
    tok_lens.append(L); pairs.append((s, ss, L))

tok_lens = np.array(tok_lens)
print(f"  SELFIES-encodable      : {len(pairs):,}")
print("  token-length percentiles: " +
      ", ".join(f"p{q}={np.percentile(tok_lens, q):.0f}" for q in (50, 90, 99)) +
      f", max={tok_lens.max()}")

kept = [(a, b) for a, b, L in pairs if L <= CFG["MAX_LEN"]]
print(f"  after MAX_LEN<={CFG['MAX_LEN']}      : {len(kept):,} "
      f"({100*len(kept)/max(1,len(pairs)):.2f}% of encodable)")

train_smiles = [a for a, _ in kept]
train_selfies = [b for _, b in kept]
if len(train_smiles) > CFG["N_TRAIN_MAX"]:
    idx = np.random.RandomState(SEED).choice(len(train_smiles),
                                             CFG["N_TRAIN_MAX"], replace=False)
    train_smiles = [train_smiles[i] for i in idx]
    train_selfies = [train_selfies[i] for i in idx]

train_set = set(train_smiles)
print(f"\U0001f9ea Final training set: {len(train_smiles):,} molecules")
pd.DataFrame({"smiles": train_smiles, "selfies": train_selfies}).to_csv(
    CSV_DIR / "training_set.csv", index=False)

CURATION = dict(raw_smiles=len(raw_smiles), parse_or_size_dropped=int(n_dropped),
                after_canon=int(n_before_dedup), after_dedup=len(canon),
                selfies_encodable=len(pairs), after_maxlen=len(kept),
                final_training=len(train_smiles),
                selfies_len_p50=float(np.percentile(tok_lens, 50)),
                selfies_len_p99=float(np.percentile(tok_lens, 99)),
                selfies_len_max=int(tok_lens.max()), source=source_name)
````

### **For Your Read, Cell 2.2: Build the vocabulary and encode every molecule**

**What you're doing:** creating the alphabet the model writes in, and translating the
whole training set into it.

**What the output tells you:** the printed *vocab size* is how many distinct chemical
"words" (atom types, ring markers, branch tokens) appear in your data, typically 30-120
for drug-like sets. A much larger vocabulary means unusual chemistry in your training
set, which is fine but harder to learn.

**Nothing to edit here** - this is plumbing. The `encode`/`decode` pair is reused
everywhere the model reads or writes molecules.

````python
#@title Cell 2.2:  Vocabulary + encoding

BOS, EOS, PAD = "[BOS]", "[EOS]", "[PAD]"

vocab_set = set()
for s in tqdm(train_selfies, desc="Vocab scan"):
    vocab_set.update(sf.split_selfies(s))
vocab = [PAD, BOS, EOS] + sorted(vocab_set)
tok2id = {t: i for i, t in enumerate(vocab)}
id2tok = {i: t for t, i in tok2id.items()}
VOCAB_SIZE = len(vocab)
SEQ_LEN    = CFG["MAX_LEN"] + 2
PAD_ID, BOS_ID, EOS_ID = tok2id[PAD], tok2id[BOS], tok2id[EOS]
print(f"\u2713 Vocab size: {VOCAB_SIZE}   |   seq len: {SEQ_LEN}")

def encode(selfies_str, max_len=SEQ_LEN):
    toks = [BOS] + list(sf.split_selfies(selfies_str)) + [EOS]
    ids = [tok2id[t] for t in toks]
    ids += [PAD_ID] * (max_len - len(ids))
    return ids[:max_len]

def decode(ids):
    out = []
    for i in ids:
        t = id2tok[int(i)]
        if t == EOS: break
        if t in (BOS, PAD): continue
        out.append(t)
    return "".join(out)

train_ids = np.array([encode(s) for s in tqdm(train_selfies, desc="Encoding")],
                     dtype=np.int64)
print(f"\u2713 Encoded tensor shape: {train_ids.shape}")
with open(MODEL_DIR / "vocab.json", "w") as f:
    json.dump({"vocab": vocab, "seq_len": SEQ_LEN}, f, indent=2)

CURATION.update(vocab_size=VOCAB_SIZE, seq_len=SEQ_LEN)
````

# **Part 3: The Generative Prior** - a transformer that learns chemistry

This is the engine. A compact GPT-style transformer decoder learns to predict the next
SELFIES token, one atom or bond at a time. After training it becomes a **prior**: a
model that produces plausible, drug-like molecules on its own. In Part 6 we take this
prior and *steer* it toward your target profile.

A transformer is used rather than an RNN because ring closures and branch points in a
molecule often span 20+ tokens, and self-attention handles those long-range
dependencies natively.

**Sampling uses a key/value cache.** Generating a molecule one token at a time without
a cache means re-running the whole prefix through the network at every step, which is
O(T²) and dominates the runtime on CPU. The cached decoder is mathematically identical,
and **Cell 3.2 proves it** by comparing cached and full-prefix logits.

### **For Your Read, Cell 3.1: Define the model**

**What you're doing:** building the neural network. Runs in a second; nothing to edit
unless you want a bigger model.

**The printed parameter count** (~0.5-2 M) is the model's capacity. That is tiny by LLM
standards but plenty for single-molecule generation on a Colab GPU.

**If you want stronger chemistry** and have GPU time: raise `D_MODEL` and `N_LAYERS` in
`CFG`. Bigger models capture subtler patterns but train slower and can overfit small
custom datasets.

````python
#@title Cell 3.1:  Transformer model definition (with KV cache)

class CausalSelfAttention(nn.Module):
    def __init__(self, d_model, n_heads, dropout):
        super().__init__()
        assert d_model % n_heads == 0
        self.n_heads, self.head_dim = n_heads, d_model // n_heads
        self.qkv  = nn.Linear(d_model, d_model * 3)
        self.proj = nn.Linear(d_model, d_model)
        self.drop = nn.Dropout(dropout)

    def _split(self, x):
        B, T, C = x.shape
        qkv = self.qkv(x).reshape(B, T, 3, self.n_heads,
                                  self.head_dim).permute(2, 0, 3, 1, 4)
        return qkv[0], qkv[1], qkv[2]

    def forward(self, x, mask):
        B, T, C = x.shape
        q, k, v = self._split(x)
        att = (q @ k.transpose(-2, -1)) / math.sqrt(self.head_dim)
        att = att.masked_fill(mask[None, None, :, :] == 0, float("-inf"))
        att = self.drop(F.softmax(att, dim=-1))
        y = (att @ v).transpose(1, 2).contiguous().view(B, T, C)
        return self.proj(y)

    def step(self, x, cache):
        """Single-token forward using cached keys/values."""
        B, T, C = x.shape
        q, k, v = self._split(x)
        if cache.get("k") is not None:
            k = torch.cat([cache["k"], k], dim=2)
            v = torch.cat([cache["v"], v], dim=2)
        cache["k"], cache["v"] = k, v
        att = F.softmax((q @ k.transpose(-2, -1)) / math.sqrt(self.head_dim), dim=-1)
        y = (att @ v).transpose(1, 2).contiguous().view(B, 1, C)
        return self.proj(y)


class TransformerBlock(nn.Module):
    def __init__(self, d_model, n_heads, dropout):
        super().__init__()
        self.ln1 = nn.LayerNorm(d_model); self.ln2 = nn.LayerNorm(d_model)
        self.attn = CausalSelfAttention(d_model, n_heads, dropout)
        self.ff = nn.Sequential(nn.Linear(d_model, 4 * d_model), nn.GELU(),
                                nn.Linear(4 * d_model, d_model), nn.Dropout(dropout))

    def forward(self, x, mask):
        x = x + self.attn(self.ln1(x), mask)
        return x + self.ff(self.ln2(x))

    def step(self, x, cache):
        x = x + self.attn.step(self.ln1(x), cache)
        return x + self.ff(self.ln2(x))


class DeNovoCompTransformer(nn.Module):
    def __init__(self, vocab_size, d_model, n_heads, n_layers, seq_len, dropout):
        super().__init__()
        self.tok_emb = nn.Embedding(vocab_size, d_model)
        self.pos_emb = nn.Embedding(seq_len, d_model)
        self.drop    = nn.Dropout(dropout)
        self.blocks  = nn.ModuleList([TransformerBlock(d_model, n_heads, dropout)
                                      for _ in range(n_layers)])
        self.ln_f = nn.LayerNorm(d_model)
        self.head = nn.Linear(d_model, vocab_size, bias=False)
        self.head.weight = self.tok_emb.weight          # weight tying
        self.seq_len = seq_len
        self.register_buffer("mask", torch.tril(torch.ones(seq_len, seq_len)))
        self.apply(self._init)

    @staticmethod
    def _init(m):
        if isinstance(m, nn.Linear):
            nn.init.normal_(m.weight, std=0.02)
            if m.bias is not None: nn.init.zeros_(m.bias)
        elif isinstance(m, nn.Embedding):
            nn.init.normal_(m.weight, std=0.02)

    def forward(self, x):
        B, T = x.shape
        pos = torch.arange(T, device=x.device)
        h = self.drop(self.tok_emb(x) + self.pos_emb(pos)[None, :, :])
        for blk in self.blocks:
            h = blk(h, self.mask[:T, :T])
        return self.head(self.ln_f(h))

    @torch.no_grad()
    def step(self, tok, pos, caches):
        h = self.tok_emb(tok) + self.pos_emb(
            torch.tensor([pos], device=tok.device))[None, :, :]
        for blk, c in zip(self.blocks, caches):
            h = blk.step(h, c)
        return self.head(self.ln_f(h))[:, -1, :]


def build_model():
    return DeNovoCompTransformer(VOCAB_SIZE, CFG["D_MODEL"], CFG["N_HEADS"],
                                 CFG["N_LAYERS"], SEQ_LEN, CFG["DROPOUT"]).to(DEVICE)

set_seed()
model = build_model()
N_PARAMS = sum(p.numel() for p in model.parameters())
print(f"\u2713 Model built  |  parameters: {N_PARAMS:,}  ({N_PARAMS/1e6:.2f}M)")
````

### **For Your Read, Cell 3.2: Prove the cached decoder is exact**

**What you're doing:** verifying that single-token decoding with a key/value cache
reproduces the full-prefix forward pass. This is a correctness check, not a formality:
if the cache were wrong, every molecule the pipeline generates would come from a
subtly different distribution than the trained model.

**What to expect:** a maximum absolute logit difference around 1e-6 or smaller, which
is floating-point noise. The cell raises an `AssertionError` if it is not.

````python
#@title Cell 3.2:  KV-cache equivalence self-test

_tst = build_model().eval()
_x = torch.randint(3, VOCAB_SIZE, (4, 12), device=DEVICE)
with torch.no_grad():
    _full = _tst(_x)
    _caches = [dict() for _ in _tst.blocks]
    _cached = torch.stack([_tst.step(_x[:, t:t+1], t, _caches) for t in range(12)],
                          dim=1)
_diff = (_full - _cached).abs().max().item()
print(f"max |full-prefix - cached| logit difference: {_diff:.3e}")
assert _diff < 1e-4, "KV cache is NOT equivalent to the full forward pass"
print("\u2713 Cached decoder verified exact.")
del _tst, _x, _full, _cached, _caches
````

# **Part 4: Train the Prior and Sample the Baseline**

Training teaches the model to generate realistic molecules. The unsteered sample that
follows is the **experimental control**: every later claim about the effect of
goal-directed optimisation is made relative to it.

### **For Your Read, Cell 4.1: Train the prior**

**What you're doing:** this is the longest cell. A few minutes on GPU for the default
settings, around 20 minutes on CPU.

**How to read the progress:** the `loss` should fall steadily and then flatten. A
smoothly decreasing curve means learning is healthy. If it plateaus high or bounces
wildly, the model is under-trained or the learning rate is off.

**Speed vs quality:** fewer `EPOCHS` = faster but rougher molecules; more = sharper but
slower. The checkpoint is saved to `models/pretrained.pt`, so you never have to retrain
within a session. This loss curve becomes panel **g** of Extended Data Fig. 1.

````python
#@title Cell 4.1:  Train the generative prior

class SelfiesDataset(Dataset):
    def __init__(self, arr): self.arr = torch.from_numpy(arr).long()
    def __len__(self): return len(self.arr)
    def __getitem__(self, i): return self.arr[i]

set_seed()
model = build_model()
dl = DataLoader(SelfiesDataset(train_ids), batch_size=CFG["BATCH_SIZE"],
                shuffle=True, num_workers=0, drop_last=True)
optim_ = torch.optim.AdamW(model.parameters(), lr=CFG["LR"], weight_decay=0.01)
total_steps = max(1, len(dl) * CFG["EPOCHS"])
warmup = min(500, max(1, total_steps // 20))

def lr_at(step):
    if step < warmup:
        return step / max(1, warmup)
    prog = (step - warmup) / max(1, total_steps - warmup)
    return 0.5 * (1 + math.cos(math.pi * prog))

losses, epoch_loss, step, running = [], [], 0, 0.0
model.train(); _t0 = time.time()
for epoch in range(CFG["EPOCHS"]):
    pbar = tqdm(dl, desc=f"epoch {epoch+1}/{CFG['EPOCHS']}")
    ep = []
    for batch in pbar:
        batch = batch.to(DEVICE, non_blocking=True)
        x, y = batch[:, :-1], batch[:, 1:]
        loss = F.cross_entropy(model(x).reshape(-1, VOCAB_SIZE), y.reshape(-1),
                               ignore_index=PAD_ID)
        for g in optim_.param_groups:
            g["lr"] = CFG["LR"] * lr_at(step)
        optim_.zero_grad(set_to_none=True)
        loss.backward()
        torch.nn.utils.clip_grad_norm_(model.parameters(), 1.0)
        optim_.step(); step += 1
        running = 0.9 * running + 0.1 * loss.item() if running else loss.item()
        losses.append(running); ep.append(loss.item())
        pbar.set_postfix(loss=f"{running:.3f}")
    epoch_loss.append(float(np.mean(ep)))

TRAIN_MINUTES = (time.time() - _t0) / 60
torch.save(model.state_dict(), MODEL_DIR / "pretrained.pt")
np.save(NPZ_DIR / "prior_loss.npy", np.array(losses))
STEPS_PER_EPOCH = len(dl)
print(f"\u2713 Prior trained in {TRAIN_MINUTES:.1f} min  |  final loss {losses[-1]:.4f}")

plt.figure(figsize=(9, 4))
plt.plot(losses, color="#3B6EA5")
plt.xlabel("Training step"); plt.ylabel("Cross-entropy (smoothed)")
plt.title("Prior training loss"); plt.tight_layout(); plt.show()
````

### **For Your Read, Cell 4.2: Sample the baseline library**

**What you're doing:** drawing `N_SAMPLES` molecules from the untuned model to
establish the control library.

**The one knob - temperature:** `TEMPERATURE` controls adventurousness. `1.0`
reproduces the training distribution. Below 1.0 gives safer, more repetitive molecules;
above 1.0 gives more novelty but more oddities. Leave it at 1.0 for the baseline.

**On how validity is measured.** It would be circular to report validity over the
collected library, because the collector already discards anything unparseable, which
forces the number to 1.0. This cell therefore measures validity *properly*, over an
independent draw of 1,000 raw token sequences, counting how many decode to a usable
molecule. That is the number quoted in the manuscript.

````python
#@title Cell 4.2:  Sample the baseline + measure true validity

@torch.no_grad()
def sample_cached(model, n, temperature=1.0):
    """Autoregressive sampling with a KV cache (verified exact in Cell 3.2)."""
    model.eval()
    caches = [dict() for _ in model.blocks]
    tok = torch.full((n, 1), BOS_ID, dtype=torch.long, device=DEVICE)
    seq = [tok]
    finished = torch.zeros(n, dtype=torch.bool, device=DEVICE)
    for t in range(SEQ_LEN - 1):
        logits = model.step(tok, t, caches) / max(1e-6, temperature)
        nxt = torch.multinomial(F.softmax(logits, dim=-1), 1)
        nxt = torch.where(finished.unsqueeze(1), torch.full_like(nxt, PAD_ID), nxt)
        seq.append(nxt)
        finished |= nxt.squeeze(-1) == EOS_ID
        if finished.all():
            break
        tok = nxt
    return torch.cat(seq, dim=1)


def ids_to_smiles(ids):
    try:
        smi = sf.decoder(decode(ids))
        if not smi: return None
        m = Chem.MolFromSmiles(smi)
        if m is None or m.GetNumHeavyAtoms() < 5: return None
        return Chem.MolToSmiles(m, canonical=True)
    except Exception:
        return None


def generate_smiles(model, n_total, temperature=1.0, batch=500, desc="sampling"):
    out, tries = [], 0
    with tqdm(total=n_total, desc=desc) as pbar:
        while len(out) < n_total and tries < 40:
            got = sample_cached(model, min(batch, max(64, n_total - len(out))),
                                temperature)
            for row in got.cpu().numpy():
                s = ids_to_smiles(row)
                if s: out.append(s)
            tries += 1
            pbar.n = min(len(out), n_total); pbar.refresh()
    return out[:n_total]


def raw_validity(model, n=1000, seed_offset=7):
    """Fraction of RAW sampled sequences that decode to a usable molecule."""
    set_seed(SEED + seed_offset)
    raw = sample_cached(model, n, CFG["TEMPERATURE"]).cpu().numpy()
    return float(sum(1 for r in raw if ids_to_smiles(r) is not None) / n)


set_seed()
baseline_smiles = generate_smiles(model, CFG["N_SAMPLES"], CFG["TEMPERATURE"],
                                  desc="Sampling baseline")
base_valid = [s for s in baseline_smiles if Chem.MolFromSmiles(s) is not None]
pd.DataFrame({"smiles": baseline_smiles}).to_csv(
    CSV_DIR / "baseline_generated.csv", index=False)

BASE_VALIDITY = raw_validity(model, n=min(1000, 10 * CFG["N_SAMPLES"]))
print(f"\u2713 Baseline library: {len(baseline_smiles):,} molecules")
print(f"  raw decode validity over an independent draw: {BASE_VALIDITY:.4f}")

PRIOR_META = dict(n_params=int(N_PARAMS), epochs=CFG["EPOCHS"], steps=int(step),
                  steps_per_epoch=STEPS_PER_EPOCH,
                  train_minutes=float(TRAIN_MINUTES),
                  epoch_mean_loss=epoch_loss,
                  initial_smoothed_loss=float(losses[0]),
                  final_smoothed_loss=float(losses[-1]),
                  raw_decode_validity=BASE_VALIDITY)
(CSV_DIR / "prior_meta.json").write_text(json.dumps(PRIOR_META, indent=2))

draw = [Chem.MolFromSmiles(s) for s in base_valid[:8]]
Draw.MolsToGridImage([m for m in draw if m], molsPerRow=4, subImgSize=(220, 200))
````

# **Part 5: ★ Your Design Specifications ★**

**This is where you design.** Instead of accepting whatever the prior produces, you
write down the molecules you *want* as target property profiles, and the model is
driven toward them by reinforcement learning. This is the principle behind REINVENT
(Olivecrona 2017; Blaschke 2020): define a scoring function that rewards molecules
matching your goals, then fine-tune the generator to maximise it.

**What is different here:** `CAMPAIGNS` is a **list**. Every specification in it is run
against the same prior and the same baseline control in one execution, and every
downstream audit compares them side by side. The manuscript reports the four campaigns
defined below.

**How a specification is written:**
- Under `properties`, each entry is one objective. `mode` sets the shape of the goal:
  - `"range"` - reward values inside `[min, max]`
  - `"max_cap"` - reward values at or below `cap`
  - `"min_floor"` - reward values at or above `floor`
  - `"maximize"` / `"minimize"` - push the value up or down between `lo` and `hi`
  - `"target"` - reward values near a single `target` (± `tol`)
  - `weight` - how much this objective counts relative to the others
- `reference_smiles` - paste a known ligand to design **analogues**; the `similarity`
  window then rewards molecules that are related but not identical (Tanimoto on Morgan
  fingerprints).
- `required_smarts` - substructures every designed molecule must contain.
- `forbidden_smarts` - substructures to eliminate.
- `include_sa` - also reward easy-to-make molecules (recommended).
- `aggregation` - `"geometric"` means *every* objective must be reasonably satisfied;
  `"arithmetic"` is more forgiving and lets objectives trade off.

**A warning that the manuscript quantifies.** Aggregation permits compensation. An
objective with a modest weight and a gently decaying penalty can be traded away
entirely while the total score rises. In the published run, molecular-weight compliance
*fell* in all four campaigns because the QED, synthetic-accessibility and aromatic-ring
terms jointly outweighed it. Part 6 reports compliance per objective so that you can
see this happening.

### **For Your Read, Cell 5.2: Compile the desirability scorer**

**What you're doing:** turning each specification into a function that scores any
molecule from 0 (misses everything) to 1 (hits everything).

**How the score is built - the desirability idea:** each objective becomes its own 0-1
*desirability* curve. A `range` objective is 1 inside the window and ramps down
outside; `max_cap` is 1 up to the cap then falls; `maximize` climbs from `lo` to `hi`;
`target` is a bell curve. The per-objective desirabilities are then combined by weight.
With `"geometric"` aggregation, a molecule that badly fails one objective gets a
near-zero total no matter how well it does elsewhere. `"arithmetic"` averages instead,
allowing trade-offs.

Graded rather than binary scoring is used because a policy-gradient learner needs a
signal that distinguishes a near miss from a gross violation; a pass/fail filter
destroys that information.

**Structural constraints are different.** They are enforced as **hard gates**: a
molecule containing a forbidden pattern, or lacking a required one, scores zero
regardless of its properties. That reflects how categorical such requirements are in
practice.

**You don't edit this cell.** It simply reads your specifications.

````python
#@title Cell 5.2:  Desirability scorer

_mfpgen = rdFingerprintGenerator.GetMorganGenerator(radius=CFG["FP_RADIUS"],
                                                    fpSize=CFG["FP_BITS"])

def morgan_fp(x):
    m = x if isinstance(x, Chem.Mol) else Chem.MolFromSmiles(x)
    return None if m is None else _mfpgen.GetFingerprint(m)

def morgan_bits(x):
    fp = morgan_fp(x)
    if fp is None: return None
    arr = np.zeros((CFG["FP_BITS"],), dtype=np.int8)
    DataStructs.ConvertToNumpyArray(fp, arr)
    return arr

def mol_property_vector(m):
    return dict(MW=Descriptors.MolWt(m), LogP=Crippen.MolLogP(m),
                TPSA=Descriptors.TPSA(m), HBD=Lipinski.NumHDonors(m),
                HBA=Lipinski.NumHAcceptors(m), RotB=Lipinski.NumRotatableBonds(m),
                AromaticRings=rdMolDescriptors.CalcNumAromaticRings(m),
                QED=QED.qed(m))

# ---- SAscore (RDKit contrib, with a heuristic fallback) ----
try:
    _sa_path = os.path.join(RDConfig.RDContribDir, "SA_Score")
    if _sa_path not in sys.path: sys.path.append(_sa_path)
    import sascorer
    SA_SOURCE = "rdkit_contrib"
    def sa_score(mol): return float(sascorer.calculateScore(mol))
    print("\u2713 SAscore loaded from RDKit contrib.")
except Exception as e:
    SA_SOURCE = "heuristic_fallback"
    print(f"\u26a0\ufe0f  SAscore contrib unavailable ({e}) \u2014 heuristic fallback.")
    def sa_score(mol):
        n_ha = mol.GetNumHeavyAtoms(); n_r = rdMolDescriptors.CalcNumRings(mol)
        n_st = rdMolDescriptors.CalcNumAtomStereoCenters(mol)
        return float(min(10.0, max(1.0, 1 + 0.05*n_ha + 0.5*n_r + 0.8*n_st)))

def bertz_score(m):
    try: return float(GraphDescriptors.BertzCT(m))
    except Exception: return float("nan")

# ---- desirability primitives (each returns 0..1) ----
def _d_range(x, lo, hi):
    margin = max(1e-6, 0.30 * (hi - lo))
    if lo <= x <= hi: return 1.0
    d = (lo - x) / margin if x < lo else (x - hi) / margin
    return float(max(0.0, 1.0 - d))

def _d_max_cap(x, cap):
    margin = max(1e-6, 0.30 * abs(cap) + 1.0)
    return 1.0 if x <= cap else float(max(0.0, 1.0 - (x - cap) / margin))

def _d_min_floor(x, floor):
    margin = max(1e-6, 0.30 * abs(floor) + 1.0)
    return 1.0 if x >= floor else float(max(0.0, 1.0 - (floor - x) / margin))

def _d_maximize(x, lo, hi):
    if x <= lo: return 0.0
    if x >= hi: return 1.0
    return float((x - lo) / (hi - lo))

def _d_minimize(x, lo, hi):
    if x <= lo: return 1.0
    if x >= hi: return 0.0
    return float((hi - x) / (hi - lo))

def _d_target(x, target, tol):
    return float(math.exp(-0.5 * ((x - target) / max(1e-6, tol)) ** 2))

def _desirability(mode, val, cfg):
    if mode == "range":     return _d_range(val, cfg["min"], cfg["max"])
    if mode == "max_cap":   return _d_max_cap(val, cfg["cap"])
    if mode == "min_floor": return _d_min_floor(val, cfg["floor"])
    if mode == "maximize":  return _d_maximize(val, cfg["lo"], cfg["hi"])
    if mode == "minimize":  return _d_minimize(val, cfg["lo"], cfg["hi"])
    if mode == "target":    return _d_target(val, cfg["target"], cfg.get("tol", 1.0))
    return 1.0


class Scorer:
    """Compiles one DESIGN_SPEC into design_score(smiles)."""

    def __init__(self, spec):
        self.spec = spec
        self.req = [Chem.MolFromSmarts(s) for s in spec.get("required_smarts", [])]
        self.frb = [Chem.MolFromSmarts(s) for s in spec.get("forbidden_smarts", [])]
        ref = spec.get("reference_smiles")
        self.ref_fp = morgan_fp(ref) if ref else None

    def objective_keys(self):
        keys = list(self.spec["properties"].keys())
        if self.ref_fp is not None: keys.append("similarity")
        if self.spec.get("include_sa", False): keys.append("SA")
        return keys

    def __call__(self, smiles, return_breakdown=False):
        m = Chem.MolFromSmiles(smiles) if smiles else None
        if m is None:
            return (0.0, {}) if return_breakdown else 0.0
        for patt in self.frb:                      # hard gate: forbidden present
            if patt is not None and m.HasSubstructMatch(patt):
                return (0.0, {"forbidden": 0.0}) if return_breakdown else 0.0
        for patt in self.req:                      # hard gate: required missing
            if patt is not None and not m.HasSubstructMatch(patt):
                return (0.0, {"required": 0.0}) if return_breakdown else 0.0

        props = mol_property_vector(m)
        ds, ws, brk = [], [], {}
        for name, c in self.spec["properties"].items():
            if name not in props: continue
            d = _desirability(c["mode"], props[name], c)
            ds.append(d); ws.append(c.get("weight", 1.0)); brk[name] = d
        if self.ref_fp is not None:
            sim = DataStructs.TanimotoSimilarity(self.ref_fp, morgan_fp(m))
            sc = self.spec["similarity"]
            d = _d_range(sim, sc["min"], sc["max"])
            ds.append(d); ws.append(sc.get("weight", 1.0)); brk["similarity"] = d
        if self.spec.get("include_sa", False):
            d = _d_target(sa_score(m), self.spec["sa_target"], tol=2.0)
            ds.append(d); ws.append(self.spec.get("sa_weight", 0.5)); brk["SA"] = d

        ds = np.clip(np.array(ds, dtype=float), 1e-6, 1.0)
        ws = np.array(ws, dtype=float); ws = ws / ws.sum()
        if self.spec.get("aggregation", "geometric") == "geometric":
            total = float(np.exp(np.sum(ws * np.log(ds))))
        else:
            total = float(np.sum(ws * ds))
        return (total, brk) if return_breakdown else total

    def in_spec(self, prop, val):
        c = self.spec["properties"][prop]; mode = c["mode"]
        if mode == "range":     return c["min"] <= val <= c["max"]
        if mode == "max_cap":   return val <= c["cap"]
        if mode == "min_floor": return val >= c["floor"]
        if mode == "maximize":  return val >= c["lo"]
        if mode == "minimize":  return val <= c["hi"]
        if mode == "target":    return abs(val - c["target"]) <= c.get("tol", 1.0)
        return True

SCORERS = {s["name"]: Scorer(s) for s in CAMPAIGNS}
for n, sc in SCORERS.items():
    print(f"   design_score[{n}]({base_valid[0][:34]:<34}) = {sc(base_valid[0]):.3f}")
````

# **Part 6: Feasibility Preview and Goal-Directed Reinforcement Learning**

For each campaign in turn this part performs three steps.

1. **Feasibility preview.** The specification is scored against the *baseline* library
   before any RL compute is spent. Three diagnostics are produced: the distribution of
   design scores, the fraction of molecules passing the hard gates, and the mean
   desirability per objective. A distribution concentrated at zero means the
   specification needs revising; a low median caused by a *gate* has a different remedy
   from one caused by unreachable property windows, which is why both are reported.
2. **Goal-directed RL.** REINFORCE with an exponential-moving-average baseline for
   variance reduction. Each round the model proposes a batch, each molecule is scored,
   and the weights are nudged to make high-scoring molecules more likely.
3. **Designed library and validation.** The steered policy is sampled under the same
   protocol as the baseline, and per-objective compliance is compared against it.

These produce **Fig. 2**, **Fig. 3** and **Extended Data Fig. 2** of the manuscript.

### **For Your Read, Cell 6.1: Run every campaign**

**What to watch:** the per-campaign print-out shows the feasibility verdict, then the
RL reward trend, then a compliance table comparing baseline with designed.

**How to read the compliance table:** a `change` column that is positive means that
objective moved into its window. **A negative value is the important one.** It means
the aggregation rule traded that objective away in exchange for gains elsewhere, and
the composite score can rise even so. Watch molecular weight in particular.

**If the reward curve is flat:** your specification is probably unreachable from this
prior. Revise the specification, or retrain on a more relevant corpus; increasing
`RL_STEPS` will not help, because RL reweights chemistry the prior can already express
and does not invent motifs absent from the training distribution.

````python
#@title Cell 6.1:  Feasibility preview + RL + designed libraries (all campaigns)

def log_prob_of_sequence(model, seq):
    model.train()
    x, y = seq[:, :-1], seq[:, 1:]
    lp = F.log_softmax(model(x), dim=-1)
    y_lp = lp.gather(-1, y.unsqueeze(-1)).squeeze(-1)
    mask = (y != PAD_ID).float()
    return (y_lp * mask).sum(dim=1)


def run_rl(scorer, steps, snapshots=(), rl_batch=None, log_every=10, tag=""):
    """Fine-tune a fresh copy of the prior. Returns (model, history, snapshots)."""
    rl_batch = rl_batch or CFG["RL_BATCH"]
    set_seed()
    m = build_model()
    m.load_state_dict(torch.load(MODEL_DIR / "pretrained.pt", map_location=DEVICE))
    opt = torch.optim.AdamW(m.parameters(), lr=CFG["RL_LR"], weight_decay=0.0)
    ema, alpha = 0.0, 0.9
    mean_hist, best_hist, snaps = [], [], {}
    for st in range(steps):
        seq = sample_cached(m, rl_batch, temperature=1.0)
        smiles = [ids_to_smiles(r) or "" for r in seq.cpu().numpy()]
        rewards = np.array([scorer(s) for s in smiles], dtype=np.float32)
        ema = alpha * ema + (1 - alpha) * rewards.mean()
        adv = torch.from_numpy(rewards - ema).float().to(DEVICE)
        loss = -(adv * log_prob_of_sequence(m, seq)).mean()
        opt.zero_grad(set_to_none=True); loss.backward()
        torch.nn.utils.clip_grad_norm_(m.parameters(), 1.0)
        opt.step()
        mean_hist.append(float(rewards.mean())); best_hist.append(float(rewards.max()))
        if (st + 1) in snapshots:
            snaps[st + 1] = {k: v.detach().clone() for k, v in m.state_dict().items()}
        if log_every and (st + 1) % log_every == 0:
            print(f"    RL{tag} {st+1}/{steps}  mean={rewards.mean():.4f}  "
                  f"max={rewards.max():.4f}")
    return m, mean_hist, best_hist, snaps


def profile_df(smiles):
    rows = []
    for s in smiles:
        m = Chem.MolFromSmiles(s)
        if m: rows.append(mol_property_vector(m))
    return pd.DataFrame(rows)


FEASIBILITY, CAMPAIGN_SUMMARY, INSPEC = {}, {}, {}
designed_smiles, designed_valid = {}, {}

for spec in CAMPAIGNS:
    name = spec["name"]; scorer = SCORERS[name]
    print(f"\n{'='*62}\n=== CAMPAIGN {name}: {spec['title']}\n{'='*62}")
    t0 = time.time()

    # ---- 1. feasibility preview -------------------------------------
    pre_scores, pre_brk = [], []
    for s in tqdm(base_valid, desc=f"Scoring baseline vs spec {name}", leave=False):
        tot, brk = scorer(s, return_breakdown=True)
        pre_scores.append(tot); pre_brk.append(brk)
    pre_scores = np.array(pre_scores)
    gate_pass = float(np.mean([not ("forbidden" in b or "required" in b)
                               for b in pre_brk]))
    obj_means = {k: (float(np.mean([b[k] for b in pre_brk if k in b]))
                     if any(k in b for b in pre_brk) else 0.0)
                 for k in scorer.objective_keys()}
    feas = dict(campaign=name, title=spec["title"],
                median=float(np.median(pre_scores)), mean=float(pre_scores.mean()),
                q95=float(np.quantile(pre_scores, 0.95)),
                maximum=float(pre_scores.max()),
                frac_above_0p5=float(np.mean(pre_scores > 0.5)),
                gate_pass_fraction=gate_pass,
                per_objective_mean_desirability=obj_means)
    FEASIBILITY[name] = feas
    np.save(NPZ_DIR / f"pre_scores_{name}.npy", pre_scores)
    (CSV_DIR / f"feasibility_{name}.json").write_text(json.dumps(feas, indent=2))
    verdict = ("reachable" if feas["q95"] > 0.3 else
               "DEMANDING \u2014 consider revising the specification")
    print(f"  feasibility: median={feas['median']:.3f}  p95={feas['q95']:.3f}  "
          f"gate-pass={gate_pass:.1%}  \u2192 {verdict}")

    # ---- 2. goal-directed RL ----------------------------------------
    m, mean_hist, best_hist, _ = run_rl(scorer, CFG["RL_STEPS"], tag=f"[{name}]")
    torch.save(m.state_dict(), MODEL_DIR / f"designed_{name}.pt")
    np.save(NPZ_DIR / f"rl_mean_{name}.npy", np.array(mean_hist))
    np.save(NPZ_DIR / f"rl_best_{name}.npy", np.array(best_hist))

    # ---- 3. designed library + validation ----------------------------
    set_seed()
    lib = generate_smiles(m, CFG["N_SAMPLES"], CFG["TEMPERATURE"],
                          desc=f"Sampling designed {name}")
    designed_smiles[name] = lib
    designed_valid[name] = [s for s in lib if Chem.MolFromSmiles(s) is not None]
    pd.DataFrame({"smiles": lib}).to_csv(
        CSV_DIR / f"designed_generated_{name}.csv", index=False)

    prof_b, prof_d = profile_df(base_valid), profile_df(designed_valid[name])
    rows = []
    for prop in [k for k in spec["properties"] if k in prof_d.columns]:
        rows.append(dict(objective=prop,
                         baseline_in_spec=float(np.mean(
                             [scorer.in_spec(prop, v) for v in prof_b[prop]])),
                         designed_in_spec=float(np.mean(
                             [scorer.in_spec(prop, v) for v in prof_d[prop]]))))
    ins = pd.DataFrame(rows)
    ins["change"] = ins["designed_in_spec"] - ins["baseline_in_spec"]
    ins.to_csv(CSV_DIR / f"target_inspec_fractions_{name}.csv", index=False)
    INSPEC[name] = ins
    print(ins.round(3).to_string(index=False))
    regressed = ins[ins["change"] < -0.01]["objective"].tolist()
    if regressed:
        print(f"  \u26a0\ufe0f  compliance REGRESSED for: {', '.join(regressed)} "
              f"\u2014 the aggregation traded these away.")

    des_scores = np.array([scorer(s) for s in designed_valid[name]])
    base_mean = float(np.mean([scorer(s) for s in base_valid]))
    pd.DataFrame({"smiles": designed_valid[name], "design_score": des_scores}) \
        .sort_values("design_score", ascending=False) \
        .to_csv(CSV_DIR / f"designed_ranked_{name}.csv", index=False)
    des_brk = [scorer(s, return_breakdown=True)[1] for s in designed_valid[name]]
    des_gate = float(np.mean([not ("forbidden" in b or "required" in b)
                              for b in des_brk]))

    summary = dict(campaign=name, title=spec["title"], baseline_mean=base_mean,
                   designed_mean=float(des_scores.mean()),
                   baseline_median=float(np.median(pre_scores)),
                   designed_median=float(np.median(des_scores)),
                   designed_max=float(des_scores.max()),
                   rl_first=mean_hist[0], rl_last=mean_hist[-1],
                   rl_best_overall=float(max(best_hist)),
                   n_designed=len(lib), n_designed_valid=len(designed_valid[name]),
                   baseline_gate_pass=gate_pass, designed_gate_pass=des_gate,
                   minutes=float((time.time() - t0) / 60))
    CAMPAIGN_SUMMARY[name] = summary
    (CSV_DIR / f"campaign_summary_{name}.json").write_text(json.dumps(summary, indent=2))
    print(f"  design score {base_mean:.3f} \u2192 {summary['designed_mean']:.3f}"
          f"   gate-pass {gate_pass:.1%} \u2192 {des_gate:.1%}"
          f"   ({summary['minutes']:.1f} min)")

print("\n\u2713 All campaigns complete.")
````

# **Part 7: ★ The RL-Intensity Sweep ★**

This part has no counterpart in most generative pipelines, and it is the one we would
ask you not to skip.

Goal-directed RL trades **compliance** against **diversity**. Pushed hard enough, the
policy concentrates on a handful of high-scoring structures: the specification is
satisfied, but the library is one structural hypothesis repeated in many guises, and if
that hypothesis fails experimentally the campaign has no fallback. Renz and colleagues
(2019) characterised this failure mode; the published DeNovoComp run reproduced it at
the pipeline's own default setting.

The sweep measures the frontier instead of guessing at it. One RL run per campaign is
performed with the policy **snapshotted** at several step counts, and a library is
sampled from each snapshot and scored for compliance, uniqueness, internal diversity,
scaffold count, alert incidence and hard-gate compliance.

**What you should expect to see.** For specifications made of graded property
objectives, most of the attainable compliance arrives in the first few steps, after
which additional optimisation is paid for almost entirely in diversity. For a
specification with a **categorical** gate (a required substructure), the picture is
different: the gate is not satisfied by gentle steering, and a higher intensity is
genuinely required, at a correspondingly higher cost in diversity.

This produces **Fig. 4** and **Extended Data Table 4** of the manuscript.

````python
#@title Cell 7.1:  Optimisation-intensity sweep

_alert_params = FilterCatalog.FilterCatalogParams()
for _cat in (FilterCatalog.FilterCatalogParams.FilterCatalogs.PAINS,
             FilterCatalog.FilterCatalogParams.FilterCatalogs.BRENK,
             FilterCatalog.FilterCatalogParams.FilterCatalogs.NIH):
    _alert_params.AddCatalog(_cat)
ALERT_CATALOG = FilterCatalog.FilterCatalog(_alert_params)


def get_scaffold(smi):
    m = Chem.MolFromSmiles(smi)
    if m is None: return None
    try:
        sc = MurckoScaffold.GetScaffoldForMol(m)
        s = Chem.MolToSmiles(sc, canonical=True) if sc else ""
        return s or None          # acyclic molecules have no framework
    except Exception:
        return None


def internal_diversity(valid, n_samples=1000):
    if len(valid) < 2: return 0.0
    pool = random.Random(SEED).sample(valid, min(n_samples, len(valid)))
    fps = [f for f in (morgan_fp(s) for s in pool) if f is not None]
    d = []
    for i in range(len(fps)):
        d.extend([1 - x for x in DataStructs.BulkTanimotoSimilarity(fps[i], fps[i+1:])])
    return float(np.mean(d)) if d else 0.0


def library_metrics(smiles, scorer, label, campaign, rl_steps):
    valid = [s for s in smiles if Chem.MolFromSmiles(s) is not None]
    scores = np.array([scorer(s) for s in valid])
    scs, sa, qed, alert, gate = [], [], [], 0, 0
    for s in valid:
        m = Chem.MolFromSmiles(s)
        sc = get_scaffold(s)
        if sc: scs.append(sc)
        sa.append(sa_score(m)); qed.append(QED.qed(m))
        if len(ALERT_CATALOG.GetMatches(m)) > 0: alert += 1
        _, brk = scorer(s, return_breakdown=True)
        if not ("forbidden" in brk or "required" in brk): gate += 1
    counts = Counter(scs); total = max(1, sum(counts.values()))
    return dict(label=label, campaign=campaign, rl_steps=rl_steps, n=len(valid),
                design_score=float(scores.mean()),
                design_median=float(np.median(scores)),
                uniqueness=len(set(valid)) / max(1, len(valid)),
                novelty=sum(1 for s in valid if s not in train_set) / max(1, len(valid)),
                int_diversity=internal_diversity(valid),
                n_scaffolds=len(counts),
                top10_share=sum(c for _, c in counts.most_common(10)) / total,
                mean_SA=float(np.mean(sa)), mean_QED=float(np.mean(qed)),
                any_alert=alert / max(1, len(valid)),
                gate_pass=gate / max(1, len(valid)))


SNAPS = sorted(CFG["SWEEP_SNAPSHOTS"])
sweep_rows = []
for spec in CAMPAIGNS:
    name = spec["name"]
    if name not in CFG["SWEEP_CAMPAIGNS"]:
        continue
    scorer = SCORERS[name]
    print(f"\n=== SWEEP CAMPAIGN {name}: {spec['title']} ===")

    m0 = library_metrics(base_valid[:CFG["SWEEP_N"]], scorer, "baseline", name, 0)
    sweep_rows.append(m0)
    print(f"  steps=0   score={m0['design_score']:.3f} uniq={m0['uniqueness']:.3f} "
          f"div={m0['int_diversity']:.3f} scaffolds={m0['n_scaffolds']}")

    _, _, _, snaps = run_rl(scorer, max(SNAPS), snapshots=SNAPS,
                            log_every=0, tag=f"[sweep {name}]")
    m = build_model()
    for s in SNAPS:
        m.load_state_dict(snaps[s])
        set_seed()
        lib = generate_smiles(m, CFG["SWEEP_N"], CFG["TEMPERATURE"],
                              desc=f"sweep {name} rl{s}")
        mm = library_metrics(lib, scorer, f"rl{s}", name, s)
        sweep_rows.append(mm)
        pd.DataFrame({"smiles": lib}).to_csv(
            CSV_DIR / f"sweep_library_{name}_rl{s}.csv", index=False)
        print(f"  steps={s:<4} score={mm['design_score']:.3f} "
              f"uniq={mm['uniqueness']:.3f} div={mm['int_diversity']:.3f} "
              f"scaffolds={mm['n_scaffolds']} alerts={mm['any_alert']:.3f} "
              f"gate={mm['gate_pass']:.3f}")

SWEEP = pd.DataFrame(sweep_rows)
SWEEP.to_csv(CSV_DIR / "intensity_sweep.csv", index=False)
print("\n" + SWEEP.round(3).to_string(index=False))

# ---- a plain-language reading of the frontier ----
print("\n--- Suggested operating point per campaign ---")
for name in CFG["SWEEP_CAMPAIGNS"]:
    sub = SWEEP[SWEEP["campaign"] == name].sort_values("rl_steps")
    base = sub[sub["rl_steps"] == 0].iloc[0]
    cand = sub[(sub["rl_steps"] > 0) &
               (sub["design_score"] > base["design_score"]) &
               (sub["int_diversity"] > 0.75 * base["int_diversity"])]
    if len(cand):
        best = cand.sort_values("design_score").iloc[-1]
        print(f"  campaign {name}: {int(best['rl_steps'])} steps "
              f"(score {best['design_score']:.3f}, diversity "
              f"{best['int_diversity']:.3f} vs {base['int_diversity']:.3f} baseline)")
    else:
        print(f"  campaign {name}: no tested intensity improved the score while "
              f"retaining 75% of baseline diversity; this specification needs "
              f"high intensity, or a revised reward.")
````

# **Part 8: The Audit Suite**

Every library produced so far, baseline and designed, is now put through the same
battery of checks. Each one is defined by the decision it informs rather than by the
number it produces.

- **Generative benchmarks** (MOSES / GuacaMol): validity, uniqueness, novelty, internal diversity
- **Distribution shift**: symmetric Kullback-Leibler divergence from the training corpus
- **Drug-likeness**: Lipinski, Veber, QED
- **Synthesizability**: SAscore, Bertz complexity, element sanity, composite
- **Structure**: Bemis-Murcko scaffolds, functional-group frequencies
- **Chemical space**: PCA and UMAP over Morgan fingerprints
- **Structural alerts**: PAINS, BRENK, NIH
- **Applicability domain**: leverage against the training fingerprint basis
- **Triage**: GO / CAUTION / NO-GO

This produces **Fig. 5**, **Extended Data Figs. 3-6** and **Extended Data Tables 2-3**.

### **For Your Read, Cell 8.1: Generative benchmarks and distribution shift**

**What each metric means for you:**
- **Validity** - measured over raw draws, as in Cell 4.2, not over the filtered library.
- **Uniqueness** - fraction that are not duplicates. **If this drops sharply for a designed library relative to baseline, the RL push was too aggressive.** Compare against Part 7.
- **Novelty** - fraction not already in the training set.
- **Internal diversity** - mean pairwise Tanimoto distance. Higher means a broader library.

**On KL divergence:** goal-directed design is *supposed* to shift the distribution, so a
larger value for a designed library is evidence that steering worked, not a defect. The
baseline value tells you how faithfully the prior reproduced the training chemistry.

````python
#@title Cell 8.1:  Benchmarks + KL divergence

SETS = {"baseline": base_valid}
SETS.update({f"designed_{n}": designed_valid[n] for n in CAMP})
SET_ORDER = ["baseline"] + [f"designed_{n}" for n in CAMP]

def uniqueness(v): return len(set(v)) / max(1, len(v))
def novelty(v, ref): return sum(1 for s in v if s not in ref) / max(1, len(v))

_bench = [dict(set="baseline", n=len(baseline_smiles), validity=BASE_VALIDITY,
               uniqueness=uniqueness(base_valid),
               novelty=novelty(base_valid, train_set),
               int_diversity=internal_diversity(base_valid))]
for n in CAMP:
    mdl = build_model()
    mdl.load_state_dict(torch.load(MODEL_DIR / f"designed_{n}.pt", map_location=DEVICE))
    v = designed_valid[n]
    _bench.append(dict(set=f"designed_{n}", n=len(designed_smiles[n]),
                       validity=raw_validity(mdl, n=min(1000, 10*CFG["N_SAMPLES"])),
                       uniqueness=uniqueness(v), novelty=novelty(v, train_set),
                       int_diversity=internal_diversity(v)))
BENCH = pd.DataFrame(_bench).set_index("set")
BENCH.to_csv(CSV_DIR / "generative_benchmarks.csv")
print(BENCH.round(4).to_string())

BASE_DESC = compute_descriptors(base_valid, "baseline descriptors", show=False)
DES_DESC = {n: compute_descriptors(designed_valid[n], f"designed {n}", show=False)
            for n in CAMP}
BASE_DESC.to_csv(CSV_DIR / "baseline_descriptors.csv", index=False)
for n in CAMP:
    DES_DESC[n].to_csv(CSV_DIR / f"designed_descriptors_{n}.csv", index=False)

def kl_descriptors(gen, train, cols=("MW", "LogP", "TPSA", "QED"), bins=40):
    out = {}
    for c in cols:
        lo = min(train[c].min(), gen[c].min()); hi = max(train[c].max(), gen[c].max())
        edges = np.linspace(lo, hi, bins + 1)
        p, _ = np.histogram(train[c], bins=edges, density=True)
        q, _ = np.histogram(gen[c], bins=edges, density=True)
        p = p + 1e-8; q = q + 1e-8; p /= p.sum(); q /= q.sum()
        out[c] = float(0.5 * (np.sum(p*np.log(p/q)) + np.sum(q*np.log(q/p))))
    return out

KL = pd.DataFrame({"baseline": kl_descriptors(BASE_DESC, train_desc),
                   **{f"designed_{n}": kl_descriptors(DES_DESC[n], train_desc)
                      for n in CAMP}})
KL.to_csv(CSV_DIR / "kl_divergence.csv")
print("\nSymmetric KL vs training distribution:")
print(KL.round(4).to_string())
````

### **For Your Read, Cell 8.2: Drug-likeness and synthesizability**

**What the flags mean:**
- **passes_ro5** - at most one Lipinski violation (MW≤500, LogP≤5, HBD≤5, HBA≤10).
- **passes_veber** - TPSA≤140 and rotatable bonds≤10.
- **SAscore** (1 easy → 10 hard) - most makeable drug-like molecules sit below ~4.5.
- **Bertz complexity** - an independent structural-complexity cross-check.
- **Element sanity** - rejects atoms outside H, C, N, O, F, P, S, Cl, Br, I.

**Because synthetic accessibility participates in the reward**, a *lower* mean SAscore
in a designed library than in the baseline is direct evidence that optimisation
favoured more preparable chemistry rather than buying property compliance at the
expense of makeability. Watch for the opposite: a required substructure can push the
generator toward more complex molecules.

**On SCScore and AiZynthFinder.** SCScore adds a reaction-corpus-derived third signal
and AiZynthFinder performs real retrosynthetic route search. Both are heavy installs
and are not enabled by default; the composite renormalises its weights over whichever
components are available. An accessibility score estimates difficulty, whereas route
search proposes a route, and the two answer different questions. The intended pattern
is to run route search only on the Part 8 shortlist.

````python
#@title Cell 8.2:  Drug-likeness + layered synthesizability

ALLOWED_ATOMS = {"H","C","N","O","F","P","S","Cl","Br","I"}

def drug_like_flags(smiles, label):
    rows = []
    for s in smiles:
        m = Chem.MolFromSmiles(s)
        if m is None: continue
        mw = Descriptors.MolWt(m); logp = Crippen.MolLogP(m)
        hbd = Lipinski.NumHDonors(m); hba = Lipinski.NumHAcceptors(m)
        tpsa = Descriptors.TPSA(m); rotb = Lipinski.NumRotatableBonds(m)
        viol = int(mw > 500) + int(logp > 5) + int(hbd > 5) + int(hba > 10)
        rows.append(dict(smiles=s, set=label, MW=mw, LogP=logp, HBD=hbd, HBA=hba,
                         TPSA=tpsa, RotB=rotb, QED=QED.qed(m), ro5_violations=viol,
                         passes_ro5=viol <= 1,
                         passes_veber=(tpsa <= 140 and rotb <= 10)))
    return pd.DataFrame(rows)

FLAGS = pd.concat([drug_like_flags(v, k) for k, v in SETS.items()], ignore_index=True)
FLAGS.to_csv(CSV_DIR / "druglike_flags.csv", index=False)
print(FLAGS.groupby("set")[["passes_ro5","passes_veber","QED","MW"]].mean().round(3)
      .reindex(SET_ORDER).to_string())

def _minmax_norm(series, invert=False, lo_q=0.05, hi_q=0.95):
    vals = pd.to_numeric(series, errors="coerce")
    lo, hi = np.nanquantile(vals, lo_q), np.nanquantile(vals, hi_q)
    if not np.isfinite(lo) or not np.isfinite(hi) or (hi - lo) < 1e-9:
        return pd.Series(np.full(len(vals), 0.5), index=series.index)
    out = ((vals - lo) / (hi - lo)).clip(0, 1)
    return (1.0 - out if invert else out).fillna(0.5)

def synth_batch(smiles, label):
    rows = []
    for s in smiles:
        m = Chem.MolFromSmiles(s)
        if m is None: continue
        sa = sa_score(m)
        rows.append(dict(smiles=s, set=label, SAscore=sa,
                         SA_norm=max(0.0, min(1.0, (10.0 - sa) / 9.0)),
                         BertzCT=bertz_score(m),
                         atoms_ok=all(a.GetSymbol() in ALLOWED_ATOMS
                                      for a in m.GetAtoms())))
    return pd.DataFrame(rows)

SYNTH = pd.concat([synth_batch(v, k) for k, v in SETS.items()], ignore_index=True)
SYNTH["Bertz_norm"] = _minmax_norm(SYNTH["BertzCT"], invert=True)

SCSCORE_AVAILABLE = False      # set True after installing SCScore to add a third signal
_parts = [SYNTH["SA_norm"].fillna(0.5).values, SYNTH["Bertz_norm"].fillna(0.5).values]
_w = [0.5, 0.3]
if SCSCORE_AVAILABLE and "SC_norm" in SYNTH.columns:
    _parts.append(SYNTH["SC_norm"].fillna(0.5).values); _w.append(0.2)
_w = np.array(_w); _w = _w / _w.sum()
SYNTH["synth_score"] = np.clip(np.vstack(_parts).T @ _w, 0, 1)
SYNTH.to_csv(CSV_DIR / "synth_composite.csv", index=False)
print("\n" + SYNTH.groupby("set")[["SAscore","BertzCT","atoms_ok","synth_score"]]
      .mean().round(3).reindex(SET_ORDER).to_string())
print("\n\u2139\ufe0f  Retrosynthetic route search (AiZynthFinder) is not run by "
      "default; apply it to tables/top_candidates_*.csv.")
````

### **For Your Read, Cell 8.3: Scaffolds and functional groups**

**What the three scaffold numbers mean:**
- **n_scaffolds** - count of unique skeletons. More = more structurally varied leads.
- **scaffold_entropy** - how evenly molecules spread across skeletons.
- **top10_share** - fraction sitting on just the 10 commonest skeletons; lower = more diverse.

**Why it matters for design:** a strong design run should still give you *options*,
many scaffolds hitting your profile rather than many near-copies of one. If
`top10_share` is high after design, ease off `RL_STEPS` and consult Part 7.

Note that acyclic molecules have no Bemis-Murcko framework and are excluded from the
scaffold counts rather than being counted as an extra "empty" scaffold.

````python
#@title Cell 8.3:  Scaffold diversity + functional groups

def scaffold_stats(smiles, name):
    scs = [x for x in (get_scaffold(s) for s in smiles) if x]
    counts = Counter(scs); total = max(1, sum(counts.values()))
    probs = np.array([c / total for c in counts.values()])
    return dict(set=name, n_valid=len(scs), n_scaffolds=len(counts),
                scaffold_entropy=float(-(probs*np.log(probs + 1e-12)).sum()),
                top10_share=sum(c for _, c in counts.most_common(10)) / total), counts

_rows, TOP_SCAF = [], {}
for label, sm in [("training", sample_train)] + [(k, v) for k, v in SETS.items()]:
    st, cnt = scaffold_stats(sm, label)
    _rows.append(st); TOP_SCAF[label] = cnt.most_common(12)
SCAF = pd.DataFrame(_rows).set_index("set")
SCAF.to_csv(CSV_DIR / "scaffold_stats.csv")
json.dump({k: v for k, v in TOP_SCAF.items()},
          open(CSV_DIR / "top_scaffolds.json", "w"), indent=2)
print(SCAF.round(3).to_string())

FGROUPS = {
    "carboxyl": "C(=O)[OH]", "amide": "C(=O)N", "ester": "C(=O)O[C;!$(C=O)]",
    "amine_1": "[NX3;H2]", "amine_2": "[NX3;H1]", "aromatic_N": "n",
    "aromatic_O": "o", "aromatic_S": "s", "halogen": "[F,Cl,Br,I]",
    "sulfonamide": "S(=O)(=O)N", "nitrile": "C#N", "aldehyde": "[CX3H1](=O)",
    "ketone": "[CX3](=O)[#6]", "ether": "[OD2]([#6])[#6]", "hydroxyl": "[OX2H]",
    "phenyl": "c1ccccc1", "pyridine": "n1ccccc1", "trifluoromethyl": "C(F)(F)F",
}
FPATTS = {k: Chem.MolFromSmarts(v) for k, v in FGROUPS.items()}

def fg_freq(smiles):
    counts = {k: 0 for k in FGROUPS}; n_ok = 0
    for s in smiles:
        m = Chem.MolFromSmiles(s)
        if m is None: continue
        n_ok += 1
        for k, pat in FPATTS.items():
            if m.HasSubstructMatch(pat): counts[k] += 1
    return {k: v / max(1, n_ok) for k, v in counts.items()}

FGFREQ = pd.DataFrame({"training": fg_freq(sample_train),
                       **{k: fg_freq(v) for k, v in SETS.items()}})
FGFREQ.to_csv(CSV_DIR / "functional_group_freq.csv")
print("\n" + FGFREQ.round(3).to_string())
````

### **For Your Read, Cell 8.4: Chemical-space mapping**

**How to read the projections, the key design read-out:**
- **Designed points inside the training cloud** → your molecules are drug-like and close to known chemistry: safer, more synthesizable, less novel.
- **Designed points drifting to their own region** → you are exploring new chemical space: more novel, but check the applicability domain in Cell 8.5 before trusting any predicted property.

Neither is "better". It depends on whether you want safe analogues or bold new
chemotypes. The map tells you which you got.

````python
#@title Cell 8.4:  PCA + UMAP projection

def fp_matrix(smiles_list, n_max=800):
    if len(smiles_list) > n_max:
        smiles_list = random.Random(SEED).sample(smiles_list, n_max)
    rows = [a for a in (morgan_bits(s) for s in smiles_list) if a is not None]
    return np.array(rows, dtype=np.int8)

_X, _lbl = [], []
for label, sm in [("training", sample_train)] + [(k, v) for k, v in SETS.items()]:
    Xi = fp_matrix(sm); _X.append(Xi); _lbl += [label] * len(Xi)
X_all = np.vstack(_X)
print(f"Fingerprint matrix: {X_all.shape}")

pca_xy = PCA(n_components=2, random_state=SEED).fit_transform(X_all.astype(np.float32))
EMB_NAME = "UMAP"
try:
    if not HAS_UMAP: raise ImportError("umap-learn not installed")
    import umap
    emb_xy = umap.UMAP(n_components=2, n_neighbors=15, min_dist=0.1,
                       random_state=SEED, metric="jaccard").fit_transform(X_all)
except Exception as e:
    print(f"\u26a0\ufe0f  UMAP unavailable ({e}) \u2014 using t-SNE.")
    from sklearn.manifold import TSNE
    emb_xy = TSNE(n_components=2, random_state=SEED,
                  perplexity=min(30, max(5, len(X_all)//4)),
                  metric="jaccard", init="random").fit_transform(X_all.astype(np.float32))
    EMB_NAME = "t-SNE"

SPACE = pd.DataFrame({"pca1": pca_xy[:,0], "pca2": pca_xy[:,1],
                      "emb1": emb_xy[:,0], "emb2": emb_xy[:,1], "set": _lbl})
SPACE.to_csv(CSV_DIR / "chemical_space.csv", index=False)
(CSV_DIR / "embedding_method.json").write_text(json.dumps({"method": EMB_NAME}))
print(f"\u2713 Projection complete ({EMB_NAME}).")
````

### **For Your Read, Cell 8.5: Structural alerts and applicability domain**

**Alerts.** Lower is better, but a molecule with an alert is not necessarily dead;
some approved drugs carry them. For early triage they are deprioritised. **If a
specific alert keeps appearing, translate it into a SMARTS and add it to
`forbidden_smarts` in Cell 5.1**, so the generator is penalised at the point of
generation rather than filtered afterwards. The published run showed alert incidence
*rising* under optimisation precisely because no alert was declared forbidden.

**Applicability domain.** Training fingerprints are compressed by PCA and each
molecule's *leverage* measures its distance from the centre of the training data in
that space. A molecule is "in domain" if its leverage is below the standard 3p/n
threshold (Netzeva 2005). Prefer in-domain molecules for anything you will act on
experimentally.

**An honest caveat the manuscript reports.** With a small basis relative to the number
of components, 3p/n is permissive and *every* molecule can fall inside, in which case
the flag contributes nothing to the triage. The cell prints a warning when that
happens. It is a property of the configuration, not a verdict on your molecules.

````python
#@title Cell 8.5:  Structural alerts + applicability domain

def alert_flags(smi):
    m = Chem.MolFromSmiles(smi)
    if m is None: return dict(pains=False, brenk=False, nih=False, any_alert=False)
    tags = [e.GetProp("FilterSet") for e in ALERT_CATALOG.GetMatches(m)]
    return dict(pains=any("PAINS" in t for t in tags),
                brenk=any("Brenk" in t for t in tags),
                nih=any("NIH" in t for t in tags), any_alert=len(tags) > 0)

ALERTS = pd.concat(
    [pd.DataFrame([dict(smiles=s, set=k, **alert_flags(s)) for s in v])
     for k, v in SETS.items()], ignore_index=True)
ALERTS.to_csv(CSV_DIR / "structural_alerts.csv", index=False)
print(ALERTS.groupby("set")[["pains","brenk","nih","any_alert"]].mean()
      .round(4).reindex(SET_ORDER).to_string())

basis_smiles = (sample_train if len(sample_train) <= CFG["AD_BASIS_MAX"]
                else random.Random(SEED).sample(sample_train, CFG["AD_BASIS_MAX"]))

def _fp_mat_keep(smiles):
    mat, kept = [], []
    for s in smiles:
        a = morgan_bits(s)
        if a is not None: mat.append(a); kept.append(s)
    X = (np.asarray(mat, dtype=np.float32) if mat
         else np.zeros((0, CFG["FP_BITS"]), dtype=np.float32))
    return X, kept

X_basis, _ = _fp_mat_keep(basis_smiles)
k_comp = int(min(CFG["AD_COMPONENTS"], max(1, X_basis.shape[0] - 1),
                 X_basis.shape[1] - 1))
ad_pca = PCA(n_components=k_comp, svd_solver="randomized", random_state=SEED)
Z_tr = ad_pca.fit_transform(X_basis).astype(np.float32)
del X_basis
ZtZ = Z_tr.T @ Z_tr
ZtZ_inv = np.linalg.pinv(ZtZ + 1e-6 * np.eye(ZtZ.shape[0], dtype=np.float32))
p_comp, n_basis = Z_tr.shape[1], Z_tr.shape[0]
H_STAR = 3.0 * p_comp / n_basis

def leverage_and_ad(smiles, label):
    X, kept = _fp_mat_keep(smiles)
    Z = ad_pca.transform(X).astype(np.float32)
    h = np.einsum("ij,jk,ik->i", Z, ZtZ_inv, Z)
    return pd.DataFrame({"smiles": kept, "set": label,
                         "leverage": h.astype(float), "in_AD": h <= H_STAR})

AD = pd.concat([leverage_and_ad(v, k) for k, v in SETS.items()], ignore_index=True)
AD.to_csv(CSV_DIR / "applicability_domain.csv", index=False)
AD_META = dict(basis=n_basis, components=p_comp, h_star=float(H_STAR))
(CSV_DIR / "ad_meta.json").write_text(json.dumps(AD_META))
print(f"\nAD basis: {n_basis} molecules, {p_comp} components, "
      f"h* = 3p/n = {H_STAR:.4f}")
print(AD.groupby("set")[["leverage","in_AD"]].mean().round(4)
      .reindex(SET_ORDER).to_string())
if AD["in_AD"].all():
    print("\n\u26a0\ufe0f  Every molecule fell inside the applicability domain, so the "
          "in-domain\n   flag contributes NO discrimination to the triage at this "
          "configuration.\n   Report this rather than treating the check as passed. A "
          "larger basis or\n   fewer components would tighten the threshold.")
````

### **For Your Read, Cell 8.6: Go / Caution / No-Go triage**

**The rules:**
- **GO**: clears all core gates - QED ≥ `GO_QED`, SAscore ≤ `GO_SA`, no structural alert, inside the applicability domain.
- **CAUTION**: clears looser thresholds with at most one alert.
- **NO-GO**: everything else. A large NO-GO fraction is normal and healthy; triage is supposed to be selective.

**What the GO label asserts, and what it does not.** It denotes computational priority
under the stated rule set. It is not a prediction of biological activity, synthetic
success or experimental validation.

**Read the counts, not just the fractions.** Fractions are over *unique* structures, so
a library whose uniqueness collapsed contributes fewer rows. It is entirely possible,
and happened in the published run, for the unsteered baseline to yield more GO
molecules than a heavily steered library.

**Next step:** `tables/top_candidates_*.csv` is exactly what the ADMET-AI and NanoTox-AI
pipelines expect as input.

````python
#@title Cell 8.6:  Merge evidence + triage

def _dedup(df): return df.drop_duplicates(subset=["smiles","set"],
                                          keep="first").reset_index(drop=True)

_f = _dedup(FLAGS[["smiles","set","QED","passes_ro5","passes_veber","MW","LogP","TPSA"]])
_s = _dedup(SYNTH[["smiles","set","SAscore","synth_score"]])
_a = _dedup(ALERTS[["smiles","set","pains","brenk","nih","any_alert"]])
_d = _dedup(AD[["smiles","set","leverage","in_AD"]])
MASTER = (_f.merge(_s, on=["smiles","set"]).merge(_a, on=["smiles","set"])
            .merge(_d, on=["smiles","set"]))
MASTER["alert_count"] = MASTER[["pains","brenk","nih"]].sum(axis=1)
MASTER["composite"] = (0.4*MASTER["QED"] + 0.4*MASTER["synth_score"]
                       + 0.1*(1 - MASTER["alert_count"]/3)
                       + 0.1*MASTER["in_AD"].astype(int))

def triage(r):
    if (r["QED"] >= CFG["GO_QED"] and r["SAscore"] <= CFG["GO_SA"]
            and not r["any_alert"] and r["in_AD"]):
        return "GO"
    if (r["QED"] >= CFG["CAUTION_QED"] and r["SAscore"] <= CFG["CAUTION_SA"]
            and r["alert_count"] <= 1):
        return "CAUTION"
    return "NO-GO"

MASTER["triage"] = MASTER.apply(triage, axis=1)
MASTER = MASTER.sort_values("composite", ascending=False).reset_index(drop=True)
MASTER.to_csv(CSV_DIR / "master_triage.csv", index=False)
print(MASTER.groupby(["set","triage"]).size().unstack(fill_value=0)
      .reindex(SET_ORDER).to_string())

_rows = []
for label in SET_ORDER:
    sub = MASTER[MASTER["set"] == label]
    go = sub[sub["triage"] == "GO"]
    _rows.append(dict(set=label, n_GO=len(go),
                      n_CAUTION=int((sub["triage"] == "CAUTION").sum()),
                      n_NOGO=int((sub["triage"] == "NO-GO").sum()),
                      n_total=len(sub),
                      go_scaffolds=len({x for x in (get_scaffold(s)
                                                    for s in go["smiles"]) if x})))
TRI_SUM = pd.DataFrame(_rows).set_index("set")
TRI_SUM.to_csv(CSV_DIR / "triage_summary.csv")
print("\n" + TRI_SUM.to_string())

for n in CAMP:
    ranked = pd.read_csv(CSV_DIR / f"designed_ranked_{n}.csv")
    g = MASTER[(MASTER["set"] == f"designed_{n}") & (MASTER["triage"] == "GO")]
    g = g.merge(ranked, on="smiles", how="left")
    g.sort_values("composite", ascending=False).head(20).to_csv(
        CSV_DIR / f"top_candidates_{n}.csv", index=False)
    if len(g) == 0:
        print(f"\u26a0\ufe0f  Campaign {n}: no molecule cleared the GO gates. "
              f"Loosen GO_QED/GO_SA, widen the spec, or lower RL intensity.")
````

# **Part 9: The Run Record**

Every number that appears in the manuscript is collected into a single
machine-readable file, `run_record.json`. The manuscript text is generated from this
record, so no reported value can diverge from the run that produced it. If you rerun
this notebook with different settings, this is the file to diff.

````python
#@title Cell 9.1:  Assemble run_record.json

def _d(df): return {k: {c: float(v) for c, v in row.items()}
                    for k, row in df.to_dict("index").items()}

RECORD = {}
RECORD["config"] = CFG
RECORD["curation"] = CURATION
RECORD["prior"] = PRIOR_META
RECORD["ad_meta"] = AD_META
RECORD["embedding"] = EMB_NAME
RECORD["sa_source"] = SA_SOURCE
RECORD["campaign_titles"] = TITLES
RECORD["specs"] = {s["name"]: {k: v for k, v in s.items() if k != "name"}
                   for s in CAMPAIGNS}
RECORD["feasibility"] = FEASIBILITY
RECORD["campaign"] = CAMPAIGN_SUMMARY
RECORD["benchmarks"] = _d(BENCH)
RECORD["kl"] = {c: {i: float(KL.loc[i, c]) for i in KL.index} for c in KL.columns}
RECORD["druglike"] = _d(FLAGS.groupby("set")[["passes_ro5","passes_veber","QED","MW",
                                              "LogP","TPSA","ro5_violations"]].mean())
RECORD["synth"] = _d(SYNTH.groupby("set")[["SAscore","BertzCT","atoms_ok",
                                           "synth_score"]].mean())
RECORD["scaffold"] = _d(SCAF)
RECORD["alerts"] = _d(ALERTS.groupby("set")[["pains","brenk","nih","any_alert"]].mean())
RECORD["ad"] = _d(AD.groupby("set")[["leverage","in_AD"]].mean())
RECORD["triage"] = _d(TRI_SUM)
RECORD["inspec"] = {n: {r["objective"]: dict(baseline=float(r["baseline_in_spec"]),
                                             designed=float(r["designed_in_spec"]),
                                             change=float(r["change"]))
                        for _, r in INSPEC[n].iterrows()} for n in CAMP}
RECORD["functional_groups"] = {c: {i: float(FGFREQ.loc[i, c]) for i in FGFREQ.index}
                               for c in FGFREQ.columns}
RECORD["sweep"] = SWEEP.round(5).to_dict("records")
RECORD["sweep_campaigns"] = sorted(SWEEP["campaign"].unique().tolist())
RECORD["sweep_steps"] = sorted(int(x) for x in SWEEP["rl_steps"].unique())
RECORD["top_candidates"] = {}
for n in CAMP:
    try:
        t = pd.read_csv(CSV_DIR / f"top_candidates_{n}.csv").head(5)
        RECORD["top_candidates"][n] = t[["smiles","QED","SAscore","synth_score",
                                         "composite","leverage"]].round(4) \
                                       .to_dict("records")
    except Exception:
        RECORD["top_candidates"][n] = []

(OUT / "run_record.json").write_text(json.dumps(RECORD, indent=2, default=float))
print(f"\u2713 run_record.json written ({(OUT/'run_record.json').stat().st_size:,} bytes)")
````
